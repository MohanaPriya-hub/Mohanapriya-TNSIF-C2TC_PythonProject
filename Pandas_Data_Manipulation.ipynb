# Pandas (DataFrames & Data Manipulation)

**Pandas** is a popular Python library used for data analysis and manipulation. It introduces two main data structures: **Series** (1D column) and **DataFrame** (2D table with rows and columns, similar to an Excel sheet).

### Key Concepts:

| Concept | Description |
| :--- | :--- |
| **Series** | A single column of data |
| **DataFrame** | A multi-column table (made of rows and columns) |
| **Filtering** | Selecting rows based on specific condition logic |
| **Grouping** | Aggregating data based on unique categories |
---

### Core Functions Reference:

| Function | Usage |
| :--- | :--- |
| `pd.Series()` | Create a 1D labeled array |
| `pd.DataFrame()` | Create a 2D labeled tabular structure |
| `pd.read_csv()` / `pd.read_excel()` | Load data from CSV or Excel files |
| `df.head()` / `df.tail()` | View first or last $n$ rows |
| `df.info()` / `df.describe()` | Get structural summary and statistical measures |
| `df['col']` / `df[['col1','col2']]` | Access single or multiple columns |
| `df.loc[]` / `df.iloc[]` | Access rows/columns by label (`loc`) or integer index (`iloc`) |
| `df.isnull()` / `df.dropna()` / `df.fillna()` | Detect, remove, or replace missing values |
| `df.drop(columns=[])` | Drop specified columns |
| `df.sort_values()` | Sort rows by values in a column |
| `df.groupby()` | Group data for aggregation |
| `df.merge()` / `df.join()` / `pd.concat()` | Combine multiple DataFrames |
| `df.to_csv()` | Save DataFrame to a CSV file |





1. Creation & Loading Data

```python
import pandas as pd

# Series: Daily bamboo intake (kg) per panda
bamboo_intake = pd.Series([12, 18, 15, 22], name="Bamboo_kg")

# DataFrame: Panda nap schedule & weight records
sanctuary_log = pd.DataFrame({
    "Bear_Name": ["Bao Bao", "Po", "Mei Xiang"],
    "Daily_Nap_Hours": [14, 16, 12],
    "Fluffiness_Score": [98, 100, 95]
})

print("Daily Bamboo Consumed (kg):\n", bamboo_intake)
print("\nSanctuary Log:\n", sanctuary_log)

# Load data from external files (syntax reference)
# sanctuary_df = pd.read_csv("panda_health_records.csv")
# sanctuary_df = pd.read_excel("panda_health_records.xlsx")
```

2. Viewing & Selecting Data (loc / iloc)

```python
import pandas as pd

data = {
    "Bear_Name": ["Po", "Bao Bao", "Xiao Qi"],
    "Favorite_Move": ["Somersault", "Tree Sticking", "Tummy Slide"],
    "Caretaker_Alert": ["High (Eating non-stop)", "Low (Asleep)", "Medium (Rolling down hill)"]
}
df = pd.DataFrame(data)

# Access single vs multiple columns
print("Favorite Moves:\n", df["Favorite_Move"])
print("\nBear & Alert Status:\n", df[["Bear_Name", "Caretaker_Alert"]])

# Access using iloc (first panda record by index)
print("\nFirst enclosure resident (iloc):\n", df.iloc[0])

# Access using loc (Name and Alert of row 0)
print("\nRow 0 details (loc):\n", df.loc[0, ["Bear_Name", "Caretaker_Alert"]])
```

3. Handling Missing Values & Dropping Columns

```python
import pandas as pd
import numpy as np

# DataFrame with missed weight checks (escaped up a tree)
data = {
    "Bear_Name": ["Po", "Bao Bao", "Xiao Qi"],
    "Weight_kg": [110, np.nan, 95],
    "Trees_Climbed_Today": [3, 8, 2]
}
df = pd.DataFrame(data)

# Check missing health stats
print("Missing Value Check:\n", df.isnull())

# Replace missing weight with average healthy estimate (100 kg)
df["Weight_kg"] = df["Weight_kg"].fillna(100)

# Drop trees climbed column
df = df.drop(columns=["Trees_Climbed_Today"])

print("\nCleaned Sanctuary Records:\n", df)
```

4. Sorting, Grouping, Combining & Saving

```python
import pandas as pd

# DataFrame 1: Enclosure bamboo consumption
df1 = pd.DataFrame({
    "Bear_ID": [1, 2, 3],
    "Bear_Name": ["Po", "Bao Bao", "Xiao Qi"],
    "Bamboo_Cost_₹": [3500, 4800, 2900]
})

# DataFrame 2: Sanctuary zone details
df2 = pd.DataFrame({
    "Bear_ID": [1, 2, 3],
    "Zone": ["Mountain Zone", "Forest Zone", "Mountain Zone"]
})

# Sort by biggest eater
sorted_eaters = df1.sort_values(by="Bamboo_Cost_₹", ascending=False)
print("Top Bamboo Eaters First:\n", sorted_eaters)

# Merge food costs with sanctuary zone
merged_df = pd.merge(df1, df2, on="Bear_ID")
print("\nMerged Enclosure Details:\n", merged_df)

# Groupby: Average bamboo spend by zone
print("\nAverage Food Cost by Zone:\n", merged_df.groupby("Zone")["Bamboo_Cost_₹"].mean())

# Save report to CSV
# merged_df.to_csv("sanctuary_budget.csv", index=False)
```
