# Python Stacks

A **Stack** is a linear data structure that follows the **LIFO (Last In, First Out)** principle. The element added last is the first one to be removed.

### Characteristics:

* **LIFO Structure:** Access is limited to the top element.
* **Push Operation:** Adding an element to the top of the stack.
* **Pop Operation:** Removing the top element from the stack.
* **Peek/Top Operation:** Viewing the top element without removing it.

---

### Core Operations:

| Operation | Description |
| :--- | :--- |
| `append()` | Push an element onto the top of the stack |
| `pop()` | Remove and return the top element |
| `[-1]` | Peek at the top element without removing it |
| `len()` | Check the total size of the stack |

Creating a Stack

```python
# Initialize an empty stack using a Python list
stack = []
```

1. append() - Push

```python
# 1.Simulating browser session navigation history
history_stack = []

# User navigates to new URLs
history_stack.append("google.com")
history_stack.append("github.com")
history_stack.append("colab.research.google.com")

print("Active browsing history:", history_stack)
```

2. [-1] - Peek

```python
# 2. View the current active webpage without navigating back
current_page = history_stack[-1]
print("Current active page:", current_page)
```

3. pop() - Remove

```python
#3. User clicks the 'Back' button
previous_page = history_stack.pop()
print("Navigated back from:", previous_page)
print("Remaining history:", history_stack)
```

4. len() - Size

```python
# 4.Check total pages stored in back-history
print("History depth:", len(history_stack))
```

# Python Queues

A **Queue** is a linear data structure that follows the **FIFO (First In, First Out)** principle. The element added first is the first one to be removed.

### Characteristics:

* **FIFO Structure:** Elements are added at the rear (enqueue) and removed from the front (dequeue).
* **Enqueue Operation:** Adding an element to the back of the queue.
* **Dequeue Operation:** Removing an element from the front of the queue.
* **Efficient Memory:** Using `collections.deque` provides fast $O(1)$ operations for both enqueue and dequeue.

---

### Core Operations (`collections.deque`):

| Operation | Description |
| :--- | :--- |
| `append()` | Enqueue: Add an element to the rear |
| `popleft()` | Dequeue: Remove and return the front element |
| `[0]` | Peek at the front element without removing it |
| `len()` | Check the total size of the queue |

Creation of Queue

```python
from collections import deque

# Initialize an empty print queue
print_queue = deque()
```

1. append() #Enqueue (Add a document to print)

```python
# 1. Send documents to the printer
print_queue.append("Resume.pdf")
print_queue.append("Project_Report.docx")
print_queue.append("Invoice.pdf")

print("Print queue:", list(print_queue))
```

2. [0] #Peek (See next document)

```python
# 2. Check which document prints next
next_doc = print_queue[0]
print("Next in line to print:", next_doc)
```

3. popleft() #Dequeue (Print and remove document)

```python
# 3. Printer finishes printing the first file
printed_doc = print_queue.popleft()

print("Printed:", printed_doc)
print("Remaining files:", list(print_queue))
```

*4*. len() #Size (Count remaining files)

```python
# 4. Check how many files are left to print
print("Documents left:", len(print_queue))
```

# Python Linked Lists

A **Linked List** is a chain of connected items called **Nodes**.

Each Node holds two things:
1. **Data:** The value itself.
2. **Next:** A pointer to the next item in the chain.

---

### Core Operations:

| Operation | Description |
| :--- | :--- |
| `Node` | Creates an individual item holding data and a pointer |
| `insert_head()` | Adds a new item to the very front of the chain |
| `display()` | Walks through the chain and prints all items |

```python
# Step 1: Create a single Employee Node
class EmployeeNode:
    def __init__(self, role):
        self.role = role
        self.reports_to = None  # Pointer to the next person in hierarchy

# Step 2: Connect the hierarchy chain
manager = EmployeeNode("Manager")
team_lead = EmployeeNode("Team Lead")
intern = EmployeeNode("Intern")

# Build the reporting chain
manager.reports_to = team_lead
team_lead.reports_to = intern

# Step 3: Traverse down the escalation path
current_person = manager
while current_person:
    print(current_person.role, end=" -> ")
    current_person = current_person.reports_to

print("End of Chain")
```
