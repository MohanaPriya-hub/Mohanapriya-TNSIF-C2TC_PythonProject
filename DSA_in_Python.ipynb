# Python Lists — DSA Theory

A **List** in Python is a **dynamic, ordered, mutable sequence** of elements.

### Characteristics:

* **Indexed:** Items are accessed via zero-based index.
* **Mutable:** You can modify, add, or remove elements.
* **Allows Duplicates:** Lists can have multiple occurrences of the same value.
* **Heterogeneous:** Can store elements of different data types.
* **Dynamic Size:** Lists can grow or shrink as needed during execution.

---

### Core Operations in Lists:

| Operation | Description |
| :--- | :--- |
| `append()` | Add an element at the end |
| `insert()` | Insert an item at a specific index |
| `remove()` | Remove the first occurrence of a value |
| `pop()` | Remove item at a given index (default last) |
| `sort()` | Sort items in ascending/descending order |
| `clear()` | Remove all items from the list |

Creating a List

```python
# Creating a list - Example
favorite_things = [
    "Sunshine",
    "A good book",
    42,
    "Coding in Python",
    True,
    3.14
]
print("Favorite Things:", favorite_things)
print("Type of Favorite Things:", type(favorite_things))
```

1. append()

```python
# append() adds an item to the end of the line (FIFO queue / playlist style)
music_queue = ["Song A", "Song B"]

# Listener queues up the next track
music_queue.append("Song C")

print("Up next:", music_queue)
# Output: ['Song A', 'Song B', 'Song C']
```

2. insert()

```python
# insert() → Inserts an element at a specified index

# insert(index, item) inserts an element at a specific position
project_team = ["Diya", "Nila", "Tara"]

# Add Kavi right into the 2nd position (index 1)
project_team.insert(1, "Kavi")

print("Project team:", project_team)
# Output: ['Diya', 'Kavi', 'Nila', 'Tara']
```

3. remove ()

```python
# remove() → Removes the first occurrence of a specified value

# remove() deletes the first matching value from the list
attendees = ["Diya", "Nila", "Tara", "Nila"]

# Removes the first occurrence of "Nila"
attendees.remove("Nila")

print("Updated attendees:", attendees)
# Output: ['Diya', 'Tara', 'Nila']
```

4. pop ()

```python
# pop() → Removes and returns element at given index (default is last)
# pop() removes and returns the element at a specific index
team = ["Anu", "Diya", "Kavi", "Nila"]

# Remove and pick the person at index 1 ("Diya")
picked = team.pop(1)

print("Picked:", picked)
print("After pop:", team)
```

5. sort ()

```python
## sort() -> Arranges items alphabetically in ascending order
names = ["Tara", "Anu", "Nila", "Diya"]

# Sort the names from A to Z
names.sort()

print("Sorted names:", names)
# Output: ['Anu', 'Diya', 'Nila', 'Tara']
```

6. clear ()

```python
# clear() removes all elements and empties the list
names = ["Anu", "Diya", "Kavi", "Tara"]

# Clear the entire list
names.clear()

print("After clear:", names)
# Output: []
```

# Python Tuples — DSA Theory

A **Tuple** in Python is an **ordered, immutable sequence** of elements.

### Characteristics:

* **Indexed:** Items are accessed via zero-based index.
* **Immutable:** Elements cannot be modified, added, or removed after creation.
* **Allows Duplicates:** Tuples can have multiple occurrences of the same value.
* **Heterogeneous:** Can store elements of different data types.
* **Fixed Size:** Memory efficient compared to lists due to immutability.

---

### Core Operations & Functions:

| Operation / Function | Description |
| :--- | :--- |
| `count()` | Count total occurrences of a value |
| `index()` | Return index of first matching value |
| `len()` | Get the total length of the tuple |
| `tuple()` | Convert another data type (like a list) into a tuple |

```python
# Create a sample tuple
# Tuple operations: Ordered and locked coffee orders
orders = ("Tea", "Coffee", "Juice", "Coffee")

# 1. count() -> Total times "Coffee" was ordered
print("Coffee count:", orders.count("Coffee"))

# 2. index() -> Position of "Juice"
print("Juice index:", orders.index("Juice"))

# 3. len() -> Total items ordered
print("Total orders:", len(orders))

# 4. tuple() -> Lock a list into a tuple
snack_list = ["Cake", "Bun"]
snacks = tuple(snack_list)
print("Snacks:", snacks)
```

# Python Dictionaries — DSA Theory

A **Dictionary** in Python is an **unordered, mutable collection** of key-value pairs.

### Characteristics:

* **Key-Value Pairs:** Stores data as `key: value` associations.
* **Unique Keys:** Keys must be unique and immutable (e.g., strings, numbers, tuples).
* **Mutable:** Values can be updated, added, or removed.
* **Fast Lookups:** Optimized for instant key-based data retrieval ($O(1)$ average time complexity).

---

### Core Operations in Dictionaries:

| Operation | Description |
| :--- | :--- |
| `dict[key]` | Access value by key |
| `get()` | Safely retrieve value without raising an error if key missing |
| `keys()` | Return a list-like view of all keys |
| `values()` | Return a list-like view of all values |
| `items()` | Return key-value pairs as tuples |
| `pop()` | Remove and return value for a specific key |

Declaration of Dictionary

```python
# Declaring a dictionary: Coffee shop item details
coffee_item = {
    "item": "Cappuccino",
    "size": "Medium",
    "price": 120,
    "in_stock": True
}

print("Item Details:", coffee_item)
print("Type:", type(coffee_item))
```

1. dict[Keys]

```python
# Accessing dictionary values using square brackets: dict[key]
movie = {
    "title": "Leo",
    "genre": "Action",
    "rating": 8.5
}

# Fetch values directly by key name
film_title = movie["title"]
film_rating = movie["rating"]

print("Title:", film_title)
print("Rating:", film_rating)
```

2. get() — Fetch value by key safely

```python
# get() safely retrieves values without crashing if a key is missing
food_order = {
    "dish": "Dosa",
    "table": 4,
    "spice": "Medium"
}

# 1. Fetch an existing key
print("Dish:", food_order.get("dish"))

# 2. Fetch a missing key (returns None instead of an error)
print("Drink:", food_order.get("drink"))

# 3. Fetch a missing key with a fallback custom default value
print("Add-on:", food_order.get("addon", "Chutney"))
```

3. keys ()

```python
# keys() retrieves all property labels in the dictionary
smart_watch = {
    "brand": "Noise",
    "battery": 85,
    "steps": 4200,
    "bluetooth": True
}

# Fetch all keys
print("Watch metrics:", smart_watch.keys())
```

4. values ()

```python
# values() retrieves only the data values stored in the dictionary
car_specs = {
    "model": "Thar",
    "fuel": "Diesel",
    "top_speed": 155,
    "sunroof": True
}

# Fetch all stored values
print("Specifications:", car_specs.values())
```

5. items ()

```python
# items() returns key-value pairs as (key, value) tuples
gym_tracker = {
    "workout": "Cardio",
    "duration_mins": 45,
    "calories_burned": 320
}

# Fetch all key-value pairs together
print("Session summary:", gym_tracker.items())
```

6. pop ()

```python
# pop() removes a specific key and returns its value
cart_item = {
    "product": "Airpods",
    "price": 14999,
    "discount": 1500
}

# Remove discount and store its value
applied_discount = cart_item.pop("discount")

print("Discount saved:", applied_discount)
print("Updated cart:", cart_item)
```

# Python Sets — DSA Theory

A **Set** in Python is an **unordered, mutable collection** of unique elements.

### Characteristics:

* **Unique Elements:** Automatically removes duplicate values.
* **Unordered:** Items do not have a defined order or index.
* **Mutable:** You can add or remove elements, but elements themselves must be immutable.
* **Fast Membership Testing:** Highly optimized for checking if an item exists ($O(1)$ average time complexity).

---

### Core Operations in Sets:

| Operation | Description |
| :--- | :--- |
| `add()` | Add an element to the set |
| `remove()` | Remove a specific element (raises error if missing) |
| `discard()` | Remove a specific element (no error if missing) |
| `union()` | Combine elements from two sets (`set1 \| set2`) |
| `intersection()` | Return common elements in both sets (`set1 & set2`) |
| `difference()` | Return elements present in first set but not second (`set1 - set2`) |

Declaring a Set

```python
# Declaring a set: Unique skill tags (duplicates are automatically dropped)
skills = {"Python", "SQL", "Git", "Python"}

print("Skills set:", skills)
print("Type:", type(skills))
```

1. add ()

```python
# add() inserts a new unique item into the set
genres = {"Sci-Fi", "Action", "Drama"}

# Add new genres to the collection
genres.add("Thriller")
genres.add("Comedy")

print("Updated genres:", genres)
```

2. remove ()

```python
# remove() deletes a specific item (raises KeyError if not found)
instruments = {"Guitar", "Piano", "Drums", "Flute"}

# Remove an instrument from the set
instruments.remove("Flute")

print("Remaining instruments:", instruments)
```

3. discard ()

```python
# discard() -> Removes an item safely without raising an error if it is missing
travel_bag = {"Passport", "Charger", "Headphones"}

travel_bag.discard("Camera")   # Not in the set, but code won't crash
travel_bag.discard("Charger")  # Successfully removes "Charger"

print("After discard:", travel_bag)
# Output: {'Passport', 'Headphones'}
```

4. Union ()

```python
#  union() -> Combines all unique elements from both sets
frontend_tools = {"React", "HTML", "CSS"}
backend_tools = {"Node", "Python", "HTML"}

# Combine all tools
all_tools = frontend_tools.union(backend_tools)

print("All tools:", all_tools)
# Output: {'Node', 'HTML', 'CSS', 'Python', 'React'}
```

5. Intersection ()

```python
# intersection() -> Returns only the common elements present in both sets
morning_orders = {"Idli", "Vada", "Poori"}
evening_orders = {"Dosa", "Vada", "Idli"}

# Items available in both morning and evening
common_menu = morning_orders.intersection(evening_orders)

print("Common items:", common_menu)
# Output: {'Idli', 'Vada'}
```

6. Diffrence ()

```python
# difference() -> Items in the required skills set that a candidate is missing
job_requirements = {"Python", "SQL", "Docker", "AWS"}
candidate_skills = {"Python", "SQL"}

# Find missing skills the candidate needs to learn
missing_skills = job_requirements.difference(candidate_skills)

print("Skills to learn:", missing_skills)
# Output: {'Docker', 'AWS'}
```
